```python
import numpy as np
arr = np.array([1, 2, 3, 4, 5])
print("Array:", arr)
print("Mean:", np.mean(arr))
print("Squared:", arr ** 2)
```

```python
import pandas as pd
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'Score': [85, 90, 95, 88]
}
df = pd.DataFrame(data)
print("\nDataFrame:\n", df)

# Filter rows
print("\nPeople older than 30:\n", df[df['Age'] > 30])
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, label="sin(x)")
plt.title("Basic Plot Example")
plt.xlabel("x values")
plt.ylabel("sin(x)")
plt.legend()
plt.show()
```
